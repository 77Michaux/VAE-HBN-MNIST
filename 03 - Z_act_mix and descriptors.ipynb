```python
latent_train_file = r"datas\latent_train.csv"
latent_test_file = r"datas\latent_test.csv"

latent_train = pd.read_csv(latent_train_file)
latent_test = pd.read_csv(latent_test_file)
latent_train.shape, latent_test.shape
```

```python
import torch
from matplotlib import pyplot as plt

D = 50
ef = len(X_TRAIN)
ep = 400
b = 1.0

rep = f"results\\VAE\\D_{D}-beta_{b}-eff_{ef}-epoch_{ep}"
net = torch.load(rep + r"\model.pt", weights_only=False)
net.train(False)
```

```python
latent_train.columns
```

```python
%matplotlib inline
from ipywidgets import interact, IntSlider, Layout


model_type = ["TRAIN", "VAL"]
idx_kept = [int(col) for col in latent_train.columns]
n = 10
lat = [latent_train, latent_test]

def display_reco(idx, i, vector_selected, vector_value, classe) :
    plt.figure(figsize = (12, 6))
    latent = np.zeros((n, 50), dtype = np.float32)
    latent[:, idx_kept] = np.array(lat[i]).astype(np.float32)[idx:idx +n]

    vector_value = float(vector_value)/100 -5
    latent[:, vector_selected] = vector_value

    
    hat_x = net.vae.Decoder(torch.from_numpy(latent)).detach().numpy()[:, 0]
    img = np.concatenate(hat_x, axis = 1)
    
    plt.imshow(img, cmap = "gray")
    plt.title(f"Obs: {idx} - Vector selected: {vector_selected}")
    plt.colorbar()
    plt.axis("off")
    plt.show()

interact(display_reco,
         idx = IntSlider(min = 0, max =  200, step = 1, value= 0, description = "Observation:", layout = Layout(width = "500px")), 
         i = IntSlider(min = 0, max = 1, step = 1, value= 0, description = "Data type:", layout = Layout(width = "500px")),
         vector_selected = IntSlider(min = 0, max = 50 -1, step = 1, value= 0, description = "vector selected:", layout = Layout(width = "500px")),
         vector_value = IntSlider(min = 0, max = 1000, step = 1, value= 300, description = "vector value:", layout = Layout(width = "500px")),
         classe = IntSlider(min = 0, max = 9, step = 1, value= 0, description = "classe:", layout = Layout(width = "500px"))
        )
```

```python
rap = rapport_de_correlation(latent_test, labels_test) *100 //1


for col in rap.columns : 
    keep = rap[col].sort_values(ascending = False)
    print(col, dict(keep[keep >= 5.0]))
```
