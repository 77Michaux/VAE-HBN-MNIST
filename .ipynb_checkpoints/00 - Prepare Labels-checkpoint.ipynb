# MNIST labels (number, MorphoMNIST, RT_MNIST)

```python
%load_ext autoreload
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import libraries.morphomnist.measure as m

# Personnal
from libraries.data_loader import load_datas as dl
```

# Classical label

```python
path = r"datas/mnist.npz"
keys = ['x_train', 'y_train', 'x_test', 'y_test']

X_train, Y_train, X_val, Y_val = dl(path, keys)


X_train = X_train.reshape((60000, 1, 28, 28))
X_val = X_val[:].reshape((10000, 1, 28, 28))


Y_train.shape, Y_train[:10]
```

# 1. Get and Save MorphoMNIST labels

```python
colnames = ['area', 'thickness', 'slant', 'width', 'height', 'large']
```

```python
%autoreload 2
img = X_train[:, 0].copy()

def fct(img) :
    return pd.Series(np.array(m.measure_image(img, threshold = 0.5, scale=4, verbose = 0)))
```

## 1.1. Train MorphoMNIST 

```python
batchsize = 1000
batch_range_train = range(0, X_train[:].shape[0] +1, batchsize)

MorphoMNIST = []
for i, batch in enumerate(batch_range_train[:-1]) :
    print(f"Epoch {i+1}/{len(batch_range_train) -1}")
    MorphoMNIST.append(pd.DataFrame(map(fct, X_train[batch:batch_range_train[i +1], 0])))

MorphoMNIST = pd.concat(MorphoMNIST)
MorphoMNIST.columns = colnames
MorphoMNIST["number"] = Y_train
MorphoMNIST
```

```python
MorphoMNIST.to_csv(Morphomnist_path, index=False)
```

```python
MorphoMNIST = pd.read_csv(Morphomnist_path, sep = ";", index_col = None)
MorphoMNIST
```

## 1.2. Validation MorphoMNIST

```python
MorphoMNIST_val_path = os.path.join(r"02 - BN_Works\DATAS", "MorphoMNIST_validation.csv")
MorphoMNIST_val_path
```

```python
batchsize = 1000
batch_range_val = range(0, X_val[:].shape[0] +1, batchsize)

MorphoMNIST_val = []
for i, batch in enumerate(batch_range_val[:-1]) :
    print(f"Epoch {i+1}/{len(batch_range_val) -1}")
    MorphoMNIST.append(pd.DataFrame(map(fct, X_train[batch:batch_range_val[i +1], 0])))

MorphoMNIST_val = pd.concat(MorphoMNIST_val)
MorphoMNIST_val.columns = colnames
MorphoMNIST_val["number"] = Y_val
MorphoMNIST_val
```

```python
MorphoMNIST_val.to_csv(MorphoMNIST_val_path, index=False)
```

```python
batchsize = 10000
batch_range_train = range(0, X_train[:].shape[0] +1, batchsize)

MorphoMNIST_RT = []
for i, batch in enumerate(batch_range_train[:-1]) :
    print(f"Epoch {i+1}/{len(batch_range_train) -1}")
    X = X_train[batch:batch_range_train[i +1], 0]
    X[X > 0] = 255
    MorphoMNIST_RT.append(pd.DataFrame(map(fct, X)))

MorphoMNIST_RT = pd.concat(MorphoMNIST_RT)
MorphoMNIST_RT.columns = colnames
MorphoMNIST_RT["number"] = Y_train
MorphoMNIST_RT
```

# RT-MNIST labels

```python
path = r"datas/rt-mnist.npz"
keys = ["mus", "stds"] # "masks", "mus", "stds"
M = None
with np.load(path) as f:
    mus, stds = [f[k] for k in keys]

mus.shape, stds.shape, mus[:10], stds[:10]
```

# Merge labels 

```python
MorphoMNIST_RT = MorphoMNIST_RT.copy() # .loc[:, ["thickness", "width", "number"]].copy()
MorphoMNIST_RT[["mu_x", "mu_y"]] = mus
MorphoMNIST_RT[["sigma_x", "sigma_y"]] = stds

MorphoMNIST_RT
```

```python
Morphomnist_rt_path = os.path.join(r"02 - BN_Works\DATAS", "continuous_morphomnist_rt.csv")
# file_path = os.path.join(path, rep_dataset, rep_version, filename)
# MorphoMNIST_RT.to_csv(Morphomnist_rt_path, index=False, sep = ";", encoding = "utf-8")
MorphoMNIST_RT = pd.read_csv(Morphomnist_rt_path, index_col=None, sep = ";")
MorphoMNIST_RT
```

```python
import scipy as sc
from scipy.stats import gaussian_kde
```

```python
# path = r"../IMG"

w, h = len(MorphoMNIST_RT.columns)//2, 2
s = 3.5
fig = plt.figure(figsize = (s*w, s*h))

for i, col in enumerate(list(MorphoMNIST_RT.columns[0:])) :
    Y = MorphoMNIST_RT[col]
    kde_y = gaussian_kde(Y)
    
    grid = np.linspace(Y.min(), Y.max(), 100) 
    p_Y = kde_y(grid)

    plt.subplot(h, w, i+1)
    plt.title(col.upper())
    plt.plot(grid, p_Y, linewidth = 2)
    plt.fill_between(
        x= grid, 
        y1= p_Y, 
        color= "b",
        alpha= 0.2)

    # plt.ylim(0, 1.0)
    # plt.grid(axis = "y")

fig.text(0.5, -0.1, 'Densities', ha='center', fontsize = 14)
fig.text(0.08, 0.5, 'Proportions', va='center', rotation='vertical', fontsize = 14)
# fig.savefig(os.path.join(path, "Morpho-MNIST - Density.pdf"), bbox_inches = 'tight')
```
