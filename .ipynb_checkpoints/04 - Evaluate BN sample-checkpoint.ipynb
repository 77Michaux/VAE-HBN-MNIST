## 2.1. $\tilde{x}$ = Decode($\tilde{z}_{act, mix}$)

```python
tilde_z = np.zeros((tilde_z_act_mix.shape[0], 50)).astype(np.float32)
tilde_z[:, tilde_z_act_mix.columns] = tilde_z_act_mix
tilde_x = net.vae.Decoder(torch.from_numpy(tilde_z)).detach().numpy()
tilde_x.shape
```

```python
%matplotlib inline
from ipywidgets import interact, IntSlider, Layout

n = 10
labels = ["None"] +list(tilde_y_disc.columns)


def display_reco(idx, label, modality) :
    plt.figure(figsize = (12, 6))
    label = labels[label]
    
    img = None
    if label == "None" :
        img = np.concatenate(tilde_x[idx:idx+n, 0], axis = 1)
    else :
        print(label, max(tilde_y_disc[label]))
        modality = min([max(tilde_y_disc[label]), modality])
        cond_idx = np.array(tilde_y_disc[tilde_y_disc[label] == modality].index) -1
        img = np.concatenate(tilde_x[cond_idx, 0][idx:idx+n], axis = 1)
    
    plt.imshow(img, cmap = "gray")
    plt.title(f"Obs: {idx} - label: {label} - modality: {modality}")
    plt.colorbar()
    plt.axis("off")
    plt.show()

interact(display_reco,
         idx = IntSlider(min = 0, max =  200, step = 1, value= 0, description = "Observation:", layout = Layout(width = "500px")),
         label = IntSlider(min = 0, max = len(labels) -1, step = 1, value= 0, description = "label:", layout = Layout(width = "500px")),
         modality = IntSlider(min = 0, max = 9, step = 1, value= 0, description = "Modality:", layout = Layout(width = "500px"))
        )
```

# 3. MorphoMNIST mesurement

```python
import libraries.morphomnist.measure as m

colnames = ['area', 'thickness', 'slant', 'width', 'height']
def get_measures(img) :
    return pd.Series(np.array(m.measure_image(img, threshold = 0.35, scale=10, verbose = 0)))
```

```python
batch_range_val
```

```python
batchsize = 100
TILDE_X = (tilde_x*255).astype(np.uint8)
batch_range_val = range(0, TILDE_X[:].shape[0] +1, batchsize)

verified_MorphoMNIST = []
for i, batch in enumerate(batch_range_val[:-1]) :
    print(f"Epoch {i+1}/{len(batch_range_val) -1}")
    df = pd.DataFrame(map(get_measures, TILDE_X[batch:batch_range_val[i +1], 0]))
    df.columns = colnames
    df.to_csv(f"datas\\MorphoMNIST\\cur_verif\\{i:04d}")
    verified_MorphoMNIST.append(df)

verified_MorphoMNIST = pd.concat(verified_MorphoMNIST)
verified_MorphoMNIST.index = np.arange(verified_MorphoMNIST.shape[0])
```

```python
MorphoMNIST_BN_path = r"datas\MorphoMNIST\MorphoMNIST_BN_67.csv"
verified_MorphoMNIST.to_csv(MorphoMNIST_BN_path, index=False)
```

```python
def to_disc(val, select) :
    for s, sel in enumerate(select) :
        if val <= sel :
            return s
    return len(select) -1

tilde_y_disc_verified = pd.DataFrame()
for label in colnames :
    tilde_y_disc_verified[label] = verified_MorphoMNIST[label].apply((lambda val : to_disc(val, intervalls.loc[1:, label])))
```

## 3.2. Distance to the good modality

```python
def dist_to_mod(measure, belief, intervall) :
    belief = int(belief)
    real_mod = int(to_disc(measure, intervall[1:]))
    # print(measure, belief, real_mod, intervall[belief], intervall[min([len(intervall)-1, belief+1])])
    if real_mod == belief :
        return 0
    if real_mod < belief :
        return intervall[belief] -measure
    return measure -intervall[belief +1]

tilde_y_disc_distance = pd.DataFrame()
for label in colnames :
    cur_df = pd.DataFrame()
    cur_df["measure"] = verified_MorphoMNIST[label]
    cur_df["belief"] = tilde_y_disc[label]
    tilde_y_disc_distance[label] = cur_df.apply((lambda val : dist_to_mod(val.measure, val.belief, list(intervalls.loc[:, label]))), axis = 1)
```

# 4. Number prediction

```python
import torch

predicter_path = r"results\Predicter\Predicter_from_z.pt"
Predicter_from_z = torch.load(predicter_path, weights_only=False) 
```

```python
from libraries.data_loader import load_datas

path = r"datas/brut/mnist.npz"
keys = ['x_train', 'y_train', 'x_test', 'y_test']

_, Y_TRAIN, _, Y_VAL = load_datas(path, keys)
```

```python
latent_train_path = r"datas\brut\latent_train.csv"
z_act_mix_train = pd.read_csv(latent_train_path)

train_loss, train_acc =  Predicter_from_z(torch.from_numpy(np.array(z_act_mix_train).astype(np.float32)), Y_TRAIN)
train_loss.detach()/z_act_mix_train.shape[0], train_acc/z_act_mix_train.shape[0]
```

```python
test_loss, test_acc =  Predicter_from_z(torch.from_numpy(np.array(z_act_mix).astype(np.float32)), Y_VAL)
test_loss.detach()/z_act_mix.shape[0], test_acc/z_act_mix.shape[0]
```
